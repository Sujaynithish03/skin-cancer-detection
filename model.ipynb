```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# Load the pre-trained model
model = load_model('skin_lesion_model.h5')

# Define the lesion type mapping
lesion_type_dict = {
    0: 'Melanocytic nevi',
    1: 'Melanoma',
    2: 'Benign keratosis-like lesions',
    3: 'Basal cell carcinoma',
    4: 'Actinic keratoses',
    5: 'Vascular lesions',
    6: 'Dermatofibroma'
}

def preprocess_image(image_path, size):
    img = cv2.imread(image_path)  # Load the image from the path
    img = cv2.resize(img, (size, size))  # Resize to 128x128
    img = img.astype(np.float32) / 255.0  # Normalize to [0, 1]
    img = np.expand_dims(img, axis=0)  # Add batch dimension
    return img

def classify_image(model, preprocessed_image):
    predictions = model.predict(preprocessed_image)
    predicted_class_idx = np.argmax(predictions, axis=1)
    predicted_class = lesion_type_dict[predicted_class_idx[0]]
    return predicted_class

image_path = 'ISIC_0033454.jpg'

preprocessed_image = preprocess_image(image_path, 128)

predicted_class = classify_image(model, preprocessed_image)

print(f"The predicted class is: {predicted_class}")
```
